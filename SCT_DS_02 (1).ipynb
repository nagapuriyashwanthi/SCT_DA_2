```python
# Step 1 - Import Libraries
import pandas as pd

# Step 2 - Load Dataset
df = pd.read_csv('Sample - Superstore.csv', encoding='utf-8', encoding_errors='ignore')
print("Shape:", df.shape)
print(df.head())
```

```python
# Check basic info
print("Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print("\nFirst 5 rows:")
df.head()
```

```python
# Check missing values
print("Missing Values in each column:")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
print("\nPercentage of missing values:")
print(df.isnull().sum() / len(df) * 100)
```

```python
# Check duplicates
print("Total duplicate rows:", df.duplicated().sum())
```

```python
# Check current data types of date columns
print("Before conversion:")
print(df['Order Date'].dtype)
print(df['Ship Date'].dtype)

# Convert to datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Confirm conversion
print("\nAfter conversion:")
print(df['Order Date'].dtype)
print(df['Ship Date'].dtype)
print("\nSample dates:")
print(df[['Order Date', 'Ship Date']].head())
```

```python
# Calculate shipping days
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

print("Shipping Days column added!")
print(df['Shipping Days'].describe())
```

```python
# Check for any weird data
print("Negative shipping days:", (df['Shipping Days'] < 0).sum())
print("Zero shipping days:", (df['Shipping Days'] == 0).sum())
```

```python
# Check shipping days distribution
print("Shipping Days Summary:")
print(df['Shipping Days'].value_counts().sort_index().head(10))

print("\nAverage shipping days:", df['Shipping Days'].mean().round(2))
print("Max shipping days:", df['Shipping Days'].max())
print("Min shipping days:", df['Shipping Days'].min())
```

```python
# Export cleaned data to new CSV file
df.to_csv('Cleaned_Superstore.csv', index=False)
print("✅ Cleaned data exported successfully!")
print("File name: Cleaned_Superstore.csv")
print("Shape:", df.shape)
```
